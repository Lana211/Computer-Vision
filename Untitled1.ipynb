```python
import random

test_dir = '/content/data/Car Brand Classification Dataset/test'
classes = os.listdir(test_dir)

for _ in range(5):
    cls = random.choice(classes)
    folder = os.path.join(test_dir, cls)
    img_name = random.choice(os.listdir(folder))
    img_path = os.path.join(folder, img_name)

    result = cls_model.predict(img_path, verbose=False)[0]
    pred_idx = result.probs.top1
    pred_conf = result.probs.top1conf.item()

    print(f"True: {cls:15s} | Predicted: {cls_model.names[pred_idx]:15s} | Conf: {pred_conf:.3f}")
```

```python
import cv2
from ultralytics import YOLO

det_model = YOLO('yolo11n.pt')
cls_model = YOLO('runs/classify/car_brand_cls/run1-4/weights/best.pt')

video_path = '/content/Vehicle Dataset Sample 2_1080p.mp4'
cap = cv2.VideoCapture(video_path)

w, h, fps = (int(cap.get(x)) for x in (cv2.CAP_PROP_FRAME_WIDTH, cv2.CAP_PROP_FRAME_HEIGHT, cv2.CAP_PROP_FPS))
out = cv2.VideoWriter('/content/output_tracked_final.mp4', cv2.VideoWriter_fourcc(*'mp4v'), fps, (w, h))

VEHICLE_CLASSES = [2]

while cap.isOpened():
    success, frame = cap.read()
    if not success:
        break

    results = det_model.track(frame, persist=True, classes=VEHICLE_CLASSES, verbose=False)

    if results[0].boxes is not None and results[0].boxes.id is not None:
        boxes = results[0].boxes.xyxy.cpu().numpy()
        track_ids = results[0].boxes.id.cpu().numpy().astype(int)

        for box, track_id in zip(boxes, track_ids):
            x1, y1, x2, y2 = map(int, box)
            crop = frame[y1:y2, x1:x2]

            if crop.size == 0:
                continue

            cls_result = cls_model.predict(crop, verbose=False)[0]
            top1_idx = cls_result.probs.top1
            brand = cls_model.names[top1_idx]
            conf = cls_result.probs.top1conf.item()

            label = f"ID:{track_id} {brand} {conf:.2f}"
            cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
            cv2.putText(frame, label, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)

    out.write(frame)

cap.release()
out.release()
print("Done:", '/content/output_tracked_final.mp4')
```
